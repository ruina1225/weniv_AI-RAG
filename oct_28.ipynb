#제로샷과 few샷 비교


```python
from openai import OpenAI
import json
client = OpenAI()

text = "배우 연기는 좋았지만 스토리가 너무 지루했어요."

# ① Zero-shot (예시 없이)
zero_prompt = f"""
너는 영화 리뷰를 분류하는 분석가야.
다음 문장을 positive, negative, neutral 중 하나로 분류해:
"{text}"
"""

# ② Few-shot (예시 3개 제공)
fewshot_prompt = f"""
너는 영화 리뷰를 분류하는 분석가야.
아래 예시를 참고해서 새로운 리뷰의 감성을 판단해.

[예시]
리뷰: "정말 감동적이고 최고의 영화였어요."
정답: positive

리뷰: "시간 낭비였어요. 다시 보고 싶지 않아요."
정답: negative

리뷰: "배우는 좋았지만 스토리가 아쉬웠어요."
정답: neutral

[질문]
리뷰: "{text}"
정답:
"""

# 두 프롬프트 비교 실행
for name, prompt in [("Zero-shot", zero_prompt), ("Few-shot", fewshot_prompt)]:
    resp = client.chat.completions.create(
        model="gpt-4o-mini",
        temperature=0.0,
        messages=[{"role":"user","content": prompt}]
    )
    print(f"\n🧩 {name} 응답:")
    print(resp.choices[0].message.content)
```

```python
from openai import OpenAI
client = OpenAI()
```

```python
SYSTEM_PROMPT = """
[Role]
너는 따뜻하고 차분한 말투의 학습 코치야.

[Goal]
사용자의 고민이나 학습 상황을 듣고,
1️⃣ 감정을 파악하고,
2️⃣ 원인을 분석하고,
3️⃣ 실행 가능한 학습 조언을 단계적으로 제시한다.

[Style]
- 한 단계씩 reasoning(생각)을 보여준 뒤 마지막에 "조언:"으로 요약해라.
- 답변은 한국어로, 문장은 짧고 따뜻하게.
- JSON, 코드블록, 불필요한 설명은 사용하지 마라.
"""
```

```python
FEWSHOT_EXAMPLES = """
[예시1]
사용자: 오늘 공부가 잘 안돼요.
챗봇:
(1) 감정 파악: 피로와 집중력 저하로 보여요.
(2) 원인 분석: 전날 수면이 부족했거나 계획이 무리했을 수 있어요.
(3) 조언: 30분만 쉬었다가 가볍게 정리부터 해보세요.

[예시2]
사용자: 수학 공식이 너무 많아서 외우기 힘들어요.
챗봇:
(1) 감정 파악: 부담감이 느껴져요.
(2) 원인 분석: 한 번에 많은 양을 외우려는 게 원인일 수 있어요.
(3) 조언: 하루에 3개씩 나눠 외우는 방법을 시도해봐요.
"""
```

```python
def study_coach_chat(user_text):
    prompt = f"""
{SYSTEM_PROMPT}

[예시 참고 - few shot]
{FEWSHOT_EXAMPLES}

[새로운 입력]
사용자 입력 : {user_text}
챗봇 :
"""
    response = client.chat.completions.create(
      model = 'gpt-4o-mini',
      temperature = 0.3,
      messages=[{'role':'user', 'content': prompt}]
    )
    return response.choices[0].message.content.strip()
```

```python
samples = [
    "오늘 집중이 너무 안돼요.",
    "영어 단어 외우기가 너무 지루해요.",
    "수학 문제를 풀면 자꾸 틀려요.",
]

for s in samples:
    print(f"👩 사용자: {s}")
    print("🤖 챗봇:", study_coach_chat(s))
    print("-"*50)
```

```python

SYSTEM_PROMPT_EN = """
You are a warm, concise English-learning coach for Korean learners.
Return ONLY a JSON object (no extra text).

Output JSON schema:
{
  "level": "A1|A2|B1|B2|C1",
  "focus": "grammar|vocabulary|fluency|pronunciation|tone",
  "feedback": ["bullet in Korean about mistakes/strengths", "..."],
  "rewrite": {
    "natural": "more natural English (1~2 sentences)",
    "simpler": "simpler English (for lower levels)"
  },
  "vocab": [
    {"phrase": "useful phrase", "meaning_ko": "간단 한국어 뜻", "example": "short example"},
    {"phrase": "useful phrase", "meaning_ko": "간단 한국어 뜻", "example": "short example"}
  ],
  "practice": "one follow-up question in English related to the user's message"
}

Rules:
- Detect an approximate CEFR level from the user's sentence and set "level".
- Choose one main "focus" that would help most now.
- Keep "feedback" in short Korean bullets (친절하고 구체적으로).
- "rewrite.natural"과 "rewrite.simpler"는 영어로만 작성.
- "vocab" 2개 제공: 실제로 쓸만한 표현, 한국어 뜻, 짧은 예문.
- "practice"는 대화를 이어가는 간단한 영어 질문 1개.
- Keep everything concise. No emojis. No extra commentary outside JSON.
"""
```

```python
FEWSHOT_EN = [
    {
        "role": "user",
        "content": "I really into baking these days, but my cake always fall down."
    },
    {
        "role": "assistant",
        "content": """{
  "level": "B1",
  "focus": "grammar",
  "feedback": [
    "be동사 누락: 'I really into' → 'I'm really into'가 자연스러워요.",
    "'fall down' 대신 'collapse'나 'sink in the middle'가 더 정확해요."
  ],
  "rewrite": {
    "natural": "I'm really into baking these days, but my cakes keep collapsing.",
    "simpler": "I like baking these days, but my cakes often sink."
  },
  "vocab": [
    {"phrase": "be into ~", "meaning_ko": "~에 푹 빠지다", "example": "I'm into jazz lately."},
    {"phrase": "keep ~ing", "meaning_ko": "자꾸/계속 ~하다", "example": "It keeps happening."}
  ],
  "practice": "What kind of cake do you want to make next?"
}"""
    },
    {
        "role": "user",
        "content": "I didn't slept well last night and I feel not good today."
    },
    {
        "role": "assistant",
        "content": """{
  "level": "A2",
  "focus": "grammar",
  "feedback": [
    "과거동사 조합 오류: 'didn't slept' → 'didn't sleep'.",
    "'feel not good'보다 'don't feel well'이 자연스러워요."
  ],
  "rewrite": {
    "natural": "I didn't sleep well last night, and I don't feel well today.",
    "simpler": "I slept badly last night, so I feel bad today."
  },
  "vocab": [
    {"phrase": "not feel well", "meaning_ko": "몸/기분이 좋지 않다", "example": "I don't feel well today."},
    {"phrase": "sleep badly", "meaning_ko": "잠을 설핏 자다", "example": "I slept badly last night."}
  ],
  "practice": "What helps you sleep better?"
}"""
    }
]
```

```python
def english_coach_chat(text, model="gpt-4o-mini", temperature=0.0, top_p=1.0, use_fewshot=True):
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT_EN}
    ]
    if use_fewshot:
        messages += FEWSHOT_EN
    messages.append({"role": "user", "content": text})

    resp = client.chat.completions.create(
        model=model,
        temperature=temperature,
        top_p=top_p,
        response_format={"type": "json_object"},
        max_tokens=300,
        messages=messages
    )
    raw = resp.choices[0].message.content.strip()
    try:
        data = json.loads(raw)
    except Exception:
        data = {
            "level": "B1",
            "focus": "fluency",
            "feedback": ["JSON 파싱 오류가 발생했어요. 문장을 조금 줄여 다시 시도해 주세요."],
            "rewrite": {"natural": text, "simpler": text},
            "vocab": [],
            "practice": "Can you rephrase your sentence in a shorter way?"
        }
    return data
```

```python
samples = [
    "I am interesting in marketing and want to improve my presentation.",
    "I went to Melborne last weekend. It was amazing but I lost my luggages.",
    "Could you correct my sentence? 'I have an meeting at 9 am tomorrow.'"
]

for s in samples:
    out = english_coach_chat(s, temperature=0.3, use_fewshot=True)
    print(f"\n👩 User: {s}")
    print(out)
    print("-"*60)
```

```python
def get_exchange_rate(currency="KRW"):
  data = yf.Ticker(f"USD{"KRW"}=X").history(period="1d")
  rate = float(data["Close"].iloc[-1])
  return json.dumps({"rate":round(rate,2), "currency": currency})
```

```python
def search_wikipedia(topic):
  # 위키피디아 요약문을 반환하는 함수 - wikipedia.summary()
  # try except -> 예외처리 구문
  # -> try는 정상적인 호출일때 오류가 없을때
  # -> except는 오류가 발생했을때
  try:
    summary = wikipedia.summary(topic, sentences=3)
    return summary
  except wikipedia.exceptions.DisambiguationError as e:
    return f"여러 주제가 있습니다.: {e.options[:3]}"
  except Exception as e:
    return f"검색 오류: {str(e)}"
```
