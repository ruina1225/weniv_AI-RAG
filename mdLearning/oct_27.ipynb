```python
# travel_bot_fixed.py
from openai import OpenAI
import os

SYSTEM_TRAVEL = r'''
너는 여행 컨시어지이자 일정 플래너야. 목표는 사용자의 일정/예산/취향에 맞는 현실적인 여행 계획을 빠르게 제안하고, 부족한 정보를 최소 질문으로 보완하는 것.

[역할]
- 지역 전반(교통·동선·관광지·음식·계절 이벤트) 전문가
- 일정 최적화(이동시간 단축, 휴무일/피크시간 회피, 백업플랜 마련)
- 예산 가이드(저예산/중간/프리미엄 대안 동시 제시)
- 안전·현지매너·기본 규정 리마인드 (비자·환전·유심/ESIM 등)

[작업 원칙]
1) 빠른 가설 제시 → 필요한 핵심만 질문(최대 5개, 객관식/선다형 선호)
2) 동선=가깝게, 교통=환승 적게, 줄서기·휴무 피하기
3) 레벨별 예산 대안(￥/₩/US$ 중 사용자가 쓴 화폐 유지)
4) 시간표는 "HH:MM"로 표기, 날짜는 "YYYY-MM-DD"로 절대날짜 사용
5) 음식점/장소는 영업시간·정기휴무·대표메뉴·예약/대기 팁을 한 줄로 요약
6) 위험·규정은 간결 고지(예: 드론 금지, 음주 규정, 자연재해 시즌)

[질문 체크리스트(부족 시)]
- 출발/도착 도시·공항:
- 여행 날짜/기간:
- 동행(인원·연령대·이동불편 여부):
- 예산 감각(저/중/고):
- 핵심 관심사(한식·카페·자연·온천·쇼핑·테마파크 등 최대 3개):
- 숙소 선호(위치·등급·체크인/아웃 시간):
- 식단/알레르기·종교 식단 제한:
- 이동수단 선호(대중교통/렌터카/택시 혼합):

[출력 형식(기본)]
1) 요약(한 줄): 이번 계획의 핵심 가치 제안
2) 일정표(일자별 표): 시간 | 장소/식당 | 할 일 | 메모
3) 동선·교통: 이동수단, 예상 소요, 예약/티켓 팁
4) 음식/카페 픽: 3곳 내외(가격대, 대표 메뉴, 예약 팁)
5) 예산 범위: 숙박/식비/교통/입장료 대략치 (총액 하이라이트)
6) 플랜 B: 우천·휴무 시 대안 2가지
7) 다음에 물을 것: 가장 중요한 추가 확정사항 3개 이내

[톤 & 언어]
- 한국어 기본, 사용자가 영어로 쓰면 영어로
- 간결하고 실행가능하게, 과장·모호어 금지

[금지/주의]
- 법률·의료·비자 확정 자문처럼 오해 소지 표현 금지("최신 규정은 공식 사이트 확인 필요" 명시)
- 존재하지 않는 가게·열차시간 단정 금지("예상/가이드"로 표시)
- 개인정보 요구 최소화

[가정]
- 타임존: Asia/Seoul
- 오늘 날짜 기준으로 "내일/이번 주말" 같은 표현은 절대날짜로 변환해 제시
'''

DEV_TRAVEL = r'''
- 불확실하면 80% 확신선에서 임시안 제시 후, 확인이 필요한 3가지를 꼬리표로 제시
- 목록은 5개 이하, 표는 5~8행 이내 유지(가독성)
- 식당/스폿은 근접성 우선 정렬(이동 30분 초과 지양)
- 예산은 "저/중/고" 3단 그레이드로 함께 표기
'''

CHECKLIST_TEMPLATE = r'''
[여행 프로필]
출발지/도착지:
여행 기간(YYYY-MM-DD ~ YYYY-MM-DD):
인원/연령:
예산 레벨: 저/중/고
관심사(최대 3):
숙소 위치/등급:
식단 제한:
이동수단 선호:
특이사항(예: 아이 동반, 차멀미 등):
'''

JSON_GUIDE = r'''
사람이 읽는 요약과 함께, 아래 JSON도 함께 반환하세요.

{
  "dates": ["YYYY-MM-DD","YYYY-MM-DD"],
  "cities": ["Seoul","Busan"],
  "party": {"adults":2,"kids":1},
  "budget_level": "mid",
  "daily_plan": [
    {
      "date":"YYYY-MM-DD",
      "items":[
        {"time":"09:30","place":"광안리","category":"sight","note":"해변 산책"},
        {"time":"12:00","place":"밀면집 A","category":"food","price_range":"₩","note":"대기 20분 예상"}
      ]
    }
  ],
  "transport": [{"from":"서울","to":"부산","mode":"KTX","eta_min":160}],
  "fallbacks": ["우천 시 B 미술관", "휴무 시 카페 C"]
}
'''

FEWSHOT_USER = '11월 부산 1박2일 추천해줘. 카페 좋아하고 회 못 먹어.'
FEWSHOT_ASSISTANT = r'''
1) 요약: 1박2일, 해변·카페 중심, 생선회 제외 로컬맛집 위주
2) 일정표
- 2025-11-08 (토)
  - 09:10 | 부산역 도착 | KTX
  - 10:00 | 감천문화마을 | 포토스팟, 언덕길 편한 신발
  - 12:00 | 돼지국밥 A | ₩ | 대기 줄 있으니 11:40 도착 권장
  - 14:00 | 영도 카페 B | 뷰카페 | 창가석 경쟁 ↑, 비창가석도 뷰 충분
  - 17:30 | 광안리 산책 | 야경 | 분수 쇼 20:00
- 2025-11-09 (일)
  - 09:30 | 민락수변공원 | 산책
  - 11:00 | 브런치 C | ₩₩ | 예약 가능
  - 13:00 | 서면 쇼핑 | 로컬 편집숍
3) 동선·교통: 부산역→감천(버스 25분), 영도(버스 15분), 광안리(지하철 35분)
4) 음식/카페 픽: 돼지국밥 A(₩, 24h, 주문 전 합체/분리 선택), 브런치 C(₩₩, 예약 추천), 카페 B(뷰·디저트 우수)
5) 예산 범위: 숙박(₩90k~150k), 식비(₩35k/인·일), 교통(KTX 왕복 약 ₩120k)
6) 플랜 B: 비 오면 미술관 D, 실내 온천 E
7) 다음 확인 필요: 출발 도시/도착 시간, 숙소 위치, KTX vs SRT 선호
'''.strip()

def run_travel_bot(client, model="gpt-4o-mini", temperature=0.5, top_p=0.9):
    print("👩‍💻 여행 질문을 입력하세요 (비면 체크리스트 폼을 먼저 드려요):")
    user_input = input("> ").strip()

    messages = [
        {"role": "system", "content": SYSTEM_TRAVEL},
        {"role": "system", "content": DEV_TRAVEL},
        {"role": "user", "content": FEWSHOT_USER},
        {"role": "assistant", "content": FEWSHOT_ASSISTANT},
    ]

    if user_input:
        messages.append({"role": "user", "content": user_input})
    else:
        messages.append({"role": "user", "content": "여행 상담을 시작하고 싶어요. 필요한 정보를 알려주세요."})
        messages.append({"role": "assistant", "content": "아래 체크리스트를 채워주세요:\n" + CHECKLIST_TEMPLATE})

    # JSON 강제 필요 시: messages.insert(1, {"role":"system","content":"[JSON 출력 가이드]\n"+JSON_GUIDE})

    res = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        top_p=top_p
    )

    print("\n🤖 챗봇의 응답:")
    print(res.choices[0].message.content)

if __name__ == "__main__":
    if not os.getenv("OPENAI_API_KEY"):
        print("⚠️ OPENAI_API_KEY가 설정되지 않았어요. 환경변수 설정 후 다시 실행하세요.")
    client = OpenAI()
    run_travel_bot(client)
```

```python
# travel_bot_interactive.py
from openai import OpenAI
import os

SYSTEM_TRAVEL = r'''(앞서 받은 SYSTEM_TRAVEL 그대로)'''
DEV_TRAVEL = r'''(앞서 받은 DEV_TRAVEL 그대로)'''
CHECKLIST_TEMPLATE = r'''(앞서 받은 CHECKLIST_TEMPLATE 그대로)'''
JSON_GUIDE = r'''(선택)'''
FEWSHOT_USER = '11월 부산 1박2일 추천해줘. 카페 좋아하고 회 못 먹어.'
FEWSHOT_ASSISTANT = r'''(앞서 받은 FEWSHOT_ASSISTANT 그대로)'''

def run_travel_bot(client, model="gpt-4o-mini", temperature=0.5, top_p=0.9):
    print("✈️ 여행 컨시어지 시작!  (종료: exit / quit / q)")

    # 대화 컨텍스트 유지(멀티턴)
    messages = [
        {"role":"system","content": SYSTEM_TRAVEL},
        {"role":"system","content": DEV_TRAVEL},
        {"role":"user","content": FEWSHOT_USER},
        {"role":"assistant","content": FEWSHOT_ASSISTANT},
    ]

    while True:
        user_input = input("\n👩‍💻 여행 질문을 입력하세요 (빈 입력 시 체크리스트 제공):\n> ").strip()

        # 종료 키워드
        if user_input.lower() in {"exit","quit","q"}:
            print("👋 종료합니다. 좋은 여행 되세요!")
            break

        # 1) 빈 입력이면 체크리스트 보여주고, 답변을 한 번 더 받기
        if not user_input:
            print("\n📝 아래 체크리스트를 채워 붙여넣어 주세요. (그냥 Enter만 치면 취소)\n")
            print("[여행 프로필]\n" + CHECKLIST_TEMPLATE.strip())
            filled = input("\n📩 체크리스트 답변을 붙여넣으세요:\n> ").strip()
            if not filled:
                print("ℹ️ 취소되었습니다. 다시 질문을 입력해 주세요.")
                continue
            user_input = "아래 체크리스트를 바탕으로 여행 계획을 제안해줘:\n" + filled

        # 2) 사용자 메시지 추가
        messages.append({"role":"user","content": user_input})

        # 3) LLM 호출
        try:
            res = client.chat.completions.create(
                model=model,
                messages=messages,
                temperature=temperature,
                top_p=top_p
            )
            reply = res.choices[0].message.content
        except Exception as e:
            print(f"❌ API 오류: {e}")
            # 오류 시 최근 user 메시지 제거(깨끗한 컨텍스트 유지)
            messages.pop()
            continue

        # 4) 출력 및 컨텍스트 업데이트
        print("\n🤖 챗봇의 응답:\n" + reply)
        messages.append({"role":"assistant","content": reply})

if __name__ == "__main__":
    if not os.getenv("OPENAI_API_KEY"):
        print("⚠️ OPENAI_API_KEY 환경변수가 설정되지 않았습니다.")
        print("   macOS/Linux:  export OPENAI_API_KEY=... ")
        print("   Windows PS:   $env:OPENAI_API_KEY=\"...\"")
    client = OpenAI()
    run_travel_bot(client)
```

```python
# 확률상 나올 수 있는 높은 확률의 단어만 가지고 조합을 하느냐 아니냐~!
# top가 1에 가까울수록 누적확률 1 / 다양한 단어의 조합
# top가 0에 가까울수록 누적확률 0 / 필요한 단어의 조합

prompt = "사과를 주제로 짧은 광고 문구를 만들어줘."
topps = [0.2, 0.5, 1.0] # 1점에 가까울수록 다양한 단어

for p in topps:
    res = client.chat.completions.create(
        model="gpt-4o-mini",
        top_p=p,
        messages=[
            {"role": "system", "content": "너는 마케팅 카피라이터야."},
            {"role": "user", "content": prompt}
        ]
    )
    print(f"\n🎯 top_p={p}\n{res.choices[0].message.content}\n")
```

##AI챗봇재설계


```python
SYSTEM_PROMPT = """
너는 따뜻하고 간결한 AI 상담 코치다.
사용자의 한두 문장을 읽고 감정을 분석한 뒤, 상황에 맞는 짧은 조언을 생성한다.
영어 명언도 한줄로 함께 추천해준다.
반드시 JSON만 출력한다. 여분의 텍스트/마크다운/설명은 금지.

출력 형식(JSON):
{
  "emotion": "positive" | "negative" | "neutral",
  "score": 0.0~1.0,            // 선택한 emotion에 대한 신뢰도(확신도)
  "message": "1~3문장, 160자 이내, 따뜻하고 실행가능한 조언, 상황에 맞는 추천 영어 명언까지 출력"
}

규칙:
- 언어: 입력이 한국어면 한국어, 아니면 입력 언어를 그대로 따라라.
- 톤: 공감 → 구체적 행동 1가지 제안(과장/판단/의학적 진단 금지).
- 길이: message는 1~2문장, 160자 이내.
- 안전: 자해·타해·위기 뉘앙스가 감지되면 즉시 전문기관/도움 요청을 권고하고, 판단/지시 대신 안전확보 중심으로 안내.
- 일관성: 모호하거나 정보가 부족하면 추정하지 말고, 가벼운 확인 질문 형태로 1문장 제시하되 JSON 형식을 유지.

라벨 지침:
- positive: 성취·기쁨·감사·희망이 중심일 때
- negative: 좌절·불안·분노·슬픔·피로 등이 지배적일 때
- neutral: 정보 전달 위주이거나 감정 방향이 뚜렷하지 않을 때

검증:
- emotion은 반드시 세 값 중 하나.
- score는 float 0.0~1.0.
- message 외의 필드는 추가하지 않는다.
"""
```

```python
def ai_advice_chat(text, model="gpt-4o-mini", temperature=0.8, top_p=1.0, desc="다양성, 창의"):
    """입력 문장에 대해 감정 + 조언 생성 (JSON 보장 + 메타 첨부)"""
    resp = client.chat.completions.create(
        model=model,
        temperature=temperature,
        top_p=top_p,
        response_format={"type": "json_object"},  # JSON만 출력하도록 강제
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": text}
        ]
    )
    raw = resp.choices[0].message.content.strip()
    try:
        data = json.loads(raw)
    except Exception as e:
        data = {"emotion": "unknown", "score": 0.0, "message": f"JSON 오류: {raw}"}

    # desc는 API 인자가 아니므로 메타로 보관(필요 없으면 이 블록 삭제)
    data["_meta"] = {"temperature": temperature, "top_p": top_p, "desc": desc}
    return data
```

```python
def run_ai_coach():
    print("🧠 AI 조언 챗봇을 시작합니다. 현재 기분이 어떠신가요? (exit 입력 시 종료)")
    while True:
        text = input("\n👩 사용자: ").strip()
        if text.lower() == "exit":
            print("👋 챗봇을 종료합니다.")
            break
        out = ai_advice_chat(text)
        label = {"positive":"🙂 긍정","negative":"🙁 부정","neutral":"😐 중립"}.get(out["emotion"], "❓")
        print(f"🤖 감정: {label} (score: {out['score']})")
        print(f"💬 조언: {out['message']}")
```

```python
run_ai_coach()
```

```python
samples = [
    "오늘 회사에서 실수를 했어요. 너무 속상해요.",
    "드디어 프로젝트를 끝냈어요! 정말 뿌듯합니다.",
    "그냥 평범한 하루였어요.",
]

emoji = {"positive":"🙂 긍정","negative":"🙁 부정","neutral":"😐 중립"}

for s in samples:
    out = ai_advice_chat(s)
    label = emoji.get(out["emotion"], "❓")
    print(f"\n👩 문장: {s}\n🤖 감정: {label} (score: {out['score']})\n💬 조언: {out['message']}")
```

```python
def compare_json_runs(text, settings, system_prompt):
    results = []
    for s in settings:
        try:
            resp = client.chat.completions.create(
                model="gpt-4o-mini",
                temperature=s["temperature"],
                top_p=s.get("top_p", 1.0),
                max_tokens=200,
                response_format={"type": "json_object"},  # JSON 일탈 방지
                messages=[
                    {"role": "system", "content": system_prompt},
                    {"role": "user", "content": text}
                ]
            )
            raw = resp.choices[0].message.content.strip()
            data = json.loads(raw)
        except Exception as e:
            data = {"emotion":"unknown","score":0.0,"message":f"JSON 오류: {str(e)}"}
        results.append({
            "desc": s.get("desc", f"T={s['temperature']},P={s.get('top_p',1.0)}"),
            "temperature": s["temperature"],
            "top_p": s.get("top_p", 1.0),
            "data": data
        })
    return results

def print_json_compare(results):
    label = {"positive":"🙂 긍정","negative":"🙁 부정","neutral":"😐 중립"}
    for r in results:
        d = r["data"]
        print(f"\n--- {r['desc']} (T={r['temperature']}, top_p={r['top_p']}) ---")
        print(f"감정: {label.get(d.get('emotion'),'❓')} (score: {d.get('score')})")
        print(f"조언: {d.get('message')}")

# 예시 실행
settings = [
    {"temperature": 0.2, "top_p": 1.0, "desc": "보수적·안정"},
    {"temperature": 0.8, "top_p": 1.0, "desc": "다양성·창의"},
]
text = "오늘 너무 피곤하고 집중이 안 돼요."
results = compare_json_runs(text, settings, SYSTEM_PROMPT)
print_json_compare(results)
```
